```python
import datetime
import os
import pandas as pd
import pandas_datareader.data as web
import requests
from dotenv import load_dotenv


load_dotenv()

USER_BCCH = os.getenv("USER_BCCH", "xxxx")
PASS_BCCH = os.getenv("PASS_BCCH", "xxx")

series_bcch = {
    "F073.TCO.PRE.HIST.M": "Tipo_Cambio",
    "F022.TPM.TIN.D001.NO.Z.M": "TPM",
    "F074.IPC.IND.Z.EP09.C.M": "IPC",
    "F032.IMC.IND.Z.Z.EP18.Z.Z.0.M": "IMACEC",
    "F019.PPB.PRE.40.M": "Precio_cobre",
    "F073.TCR.IND.199101.M": "TCR",
    "F051.H121.STO.H.C.CLP.M": "Reservas_internacionales",
    "F049.DES.TAS.INE9.10.M": "Desempleo",
}


def get_bcch_data(serie, user, password, start_date="2013-01-01"):
    url = f"https://si3.bcentral.cl/SieteRestWS/SieteRestWS.ashx?user={user}&pass={password}&firstdate={start_date}&timeseries={serie}&function=GetSeries"
    try:
        response = requests.get(url, timeout=10)
        if response.status_code == 200:
            data = response.json()["Series"]["Obs"]
            df = pd.DataFrame(data)
            df["indexDateString"] = pd.to_datetime(
                df["indexDateString"], format="%d-%m-%Y"
            )
            df.set_index("indexDateString", inplace=True)
            df.rename(columns={"value": serie}, inplace=True)
            df[serie] = pd.to_numeric(df[serie], errors="coerce")
            return df[[serie]]
        else:
            return pd.DataFrame()
    except Exception:
        return pd.DataFrame()


# Descargar y unir datos del Banco Central de Chile
df_chile = pd.DataFrame()
for codigo, nombre in series_bcch.items():
    df_temp = get_bcch_data(codigo, USER_BCCH, PASS_BCCH, "2013-01-01")
    if not df_temp.empty:
        df_temp.rename(columns={codigo: nombre}, inplace=True)
        if df_chile.empty:
            df_chile = df_temp
        else:
            df_chile = df_chile.join(df_temp, how="outer")

# ----------------------------------------------------
# 2. EXTRACCIÓN DESDE FRED (EE.UU.)
# ----------------------------------------------------
start = datetime.datetime(2013, 1, 1)
end = datetime.datetime.now()

series_fred = {
    "FEDFUNDS": "Tasa_FED_EEUU",
    "CPIAUCSL": "IPC_EEUU",
    "DTWEXBGS": "Indice_Dolar",
    "VIXCLS": "Indice_VIX",
}

df_usa = pd.DataFrame()
for codigo, nombre in series_fred.items():
    try:
        df_temp = web.DataReader(codigo, "fred", start, end)
        df_temp.rename(columns={codigo: nombre}, inplace=True)
        if df_usa.empty:
            df_usa = df_temp
        else:
            df_usa = df_usa.join(df_temp, how="outer")
    except Exception:
        pass

# ----------------------------------------------------
# 3. CONSOLIDACIÓN DE DATOS
# ----------------------------------------------------
df_final = df_chile.join(df_usa, how="outer")
df_mensual = df_final.resample("ME").mean()

print(df_mensual.tail())
```

```python
import pandas as pd

# Mapeamos los nombres exactos de las columnas en df_mensual
info_series = {
    "Tipo_Cambio": ("F073.TCO.PRE.HIST.M", "Banco Central", "Sector Externo"),
    "TPM": ("F022.TPM.TIN.D001.NO.Z.M", "Banco Central", "Politica monetaria"),
    "IPC": ("F074.IPC.IND.Z.EP09.C.M", "Banco Central", "Precios"),
    "IMACEC": ("F032.IMC.IND.Z.Z.EP18.Z.Z.0.M", "Banco Central", "Actividad"),
    "Precio_cobre": ("F019.PPB.PRE.40.M", "Banco Central", "Sector externo"),
    "TCR": ("F073.TCR.IND.199101.M", "Banco Central", "Sector externo"),
    "Reservas_internacionales": (
        "F051.H121.STO.H.C.CLP.M",
        "Banco Central",
        "Sector externo",
    ),
    "Desempleo": ("F049.DES.TAS.INE9.10.M", "Banco Central", "Mercado laboral"),
    "Tasa_FED_EEUU": ("FEDFUNDS", "FRED", "Internacional"),
    "IPC_EEUU": ("CPIAUCSL", "FRED", "Internacional"),
    "Indice_Dolar": ("DTWEXBGS", "FRED", "Internacional"),
    "Indice_VIX": ("VIXCLS", "FRED", "Internacional"),
}

# Convertimos el df_mensual a formato largo SQLite

lista_obs = []
for col in info_series:
    if col in df_mensual.columns:
        cod_id = info_series[col][0]
        temp = df_mensual[[col]].reset_index()
        temp.columns = ["fecha", "valor"]
        temp["id_serie"] = cod_id
        lista_obs.append(temp[["id_serie", "fecha", "valor"]])

# Unimos todo en un solo DataFrame
df_obs = pd.concat(lista_obs, ignore_index=True)
df_obs = df_obs.dropna()

print("✅ Tabla de observaciones creada exitosamente:")
print(df_obs.head())


```
